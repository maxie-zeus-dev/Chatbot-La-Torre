```python
# Chatbot para Asistencia Condominal - La Torre

# Prototipo de asistente conversacional basado en reglas, desarrollado en Python y NLTK.

# Objetivo: proporcionar orientación sobre reglamentos, áreas comunes y procedimientos del condominio La Torre.

# Maestría en Inteligencia Artificial
# Módulo: Desarrollo de Aplicaciones con Inteligencia Artificial


import nltk

# Descargar recursos para tokenizar texto
nltk.download("punkt")
nltk.download("punkt_tab")


# Genero la función para poder normalizar el texto recibido
import unicodedata
from nltk.tokenize import word_tokenize
import re

def normalizar_texto(mensaje):

    # Convertir a minúsculas
    mensaje = mensaje.lower()

    # Eliminar acentos
    mensaje = ''.join(
        caracter for caracter in unicodedata.normalize("NFD", mensaje)
        if unicodedata.category(caracter) != "Mn")

    # Sustituir signos de puntuación por espacios
    mensaje = re.sub(r"[^\w\s]", " ", mensaje)

    # Separar el mensaje en tokens
    tokens = word_tokenize(mensaje, language="spanish")

    # Conservar únicamente palabras y números
    tokens = [
        token for token in tokens
        if token.isalnum()
    ]

    return tokens
```

```python
class DetectIntent:

    def __init__(self):

        # Diccionario de intenciones y patrones asociados
        self.intenciones = {
            "consultar_horarios": [
                ["horario", "alberca"],
                ["horario", "terraza"],
                ["horario"],
                ["hora", "abre"],
                ["hora", "cierra"],
                ["lunes", "alberca"]
            ],

            "consultar_uso_areas": [
                ["invitados", "alberca"],
                ["usar", "alberca"],
                ["usar", "terraza"],
                ["asadores"],
                ["asador"]
            ],

            "consultar_reservaciones": [
                ["reservar"],
                ["reservacion"],
                ["apartar"],
                ["rentar", "terraza"],
                ["tener", "evento"],
                ["fiesta"],
                ["evento"],
                ["celebracion"],
                ["carne", "asada"]
            ],

            "consultar_cuotas": [
                ["cuotas"],
                ["mantenimiento", "pagar"],
                ["pago", "mantenimiento"],
                ["adeudo"]
            ],

            "consultar_construccion": [
                ["construir"],
                ["construccion"],
                ["obra"],
                ["albaniles"],
                ["remodelacion"],
                ["horario", "construir"],
                ["horario", "construccion"],
                ["hora", "albaniles"]
            ],

            "consultar_mascotas": [
                ["mascotas"],
                ["perro"],
                ["gato"]
            ],

            "consultar_vialidad": [
                ["estacionamiento"],
                ["estacionarme"],
                ["velocidad"],
                ["vialidad"],
                ["tag"]
            ],

            "reportar_incidencia": [
                ["reportar"],
                ["queja"],
                ["denunciar"],
                ["ruido", "vecino"],
                ["incidencia"]
            ]
        }

    def detectar_intencion(self, mensaje):

        tokens = normalizar_texto(mensaje)
        conjunto_tokens = set(tokens)

        mejor_intencion = "intencion_desconocida"
        mejor_puntuacion = 0

        for intencion, patrones in self.intenciones.items():

            for patron in patrones:

                # Verificar si todas las palabras del patrón aparecen
                if all(palabra in conjunto_tokens for palabra in patron):

                    puntuacion = len(patron)

                    if puntuacion > mejor_puntuacion:
                        mejor_puntuacion = puntuacion
                        mejor_intencion = intencion

        return mejor_intencion
```

```python
class EntityExtractor:

    def __init__(self):

        self.areas = [
            "alberca",
            "terraza",
            "jardin",
            "asador"
        ]

        self.dias = [
            "lunes", "martes", "miercoles",
            "jueves", "viernes", "sabado", "domingo"
        ]

    def extraer_entidades(self, mensaje):

        tokens = normalizar_texto(mensaje)
        palabras = set(tokens)
        entidades = {
            "area": None,
            "dia": None,
            "cantidad": None
        }

        # Identificar el área común
        for area in self.areas:
            if area in palabras:
                entidades["area"] = area
                break

        # Identificar el día de la semana
        for dia in self.dias:
            if dia in palabras:
                entidades["dia"] = dia
                break

        # Identificar cantidades numéricas
        for token in tokens:
            if token.isdigit():
                entidades["cantidad"] = int(token)
                break

        return entidades
```

```python

class ConversationManager:

    def __init__(self):

        # Crear los componentes necesarios
        self.detector = DetectIntent()
        self.base = KnowledgeBase()
        self.extractor = EntityExtractor()
        self.estado = None

    def identificar_tema(self, intencion, mensaje):

        tokens = normalizar_texto(mensaje)
        palabras = set(tokens)

        if intencion == "consultar_horarios":

            if "alberca" in palabras:
                return "alberca"

            elif "terraza" in palabras:
                return "terraza"

        elif intencion == "consultar_uso_areas":

            if "alberca" in palabras and "invitados" in palabras:
                return "invitados_alberca"

        elif intencion == "consultar_reservaciones":
            return "terraza"

        elif intencion == "consultar_cuotas":
            # Identificar preguntas sobre fechas de pago
            if any(palabra in palabras for palabra in
                  ["cuando", "fecha", "vence", "vencimiento"]):
                  return "fecha_pago"

            # Identificar preguntas sobre responsabilidad de pago
            elif any(palabra in palabras for palabra in
                  ["quien", "responsable", "obligacion"]):
                  return "mantenimiento"

            return None

        elif intencion == "consultar_construccion":
            return "horarios_obra"

        elif intencion == "consultar_mascotas":

            if "terraza" in palabras:
                return "terraza"

        elif intencion == "consultar_vialidad":

            if "velocidad" in palabras:
                return "velocidad"

            elif (
                "estacionamiento" in palabras
                or "estacionarme" in palabras
            ):
                return "estacionamiento_visitas"

        elif intencion == "reportar_incidencia":
            return "queja"

        return None

    def evaluar_restricciones(self, intencion, mensaje):

        entidades = self.extractor.extraer_entidades(mensaje)
        palabras = set(normalizar_texto(mensaje))

        restricciones = []

        # Reglas específicas de la alberca
        if entidades["area"] == "alberca":

            if entidades["dia"] == "lunes":
                restricciones.append(
                    "La alberca no puede utilizarse los lunes "
                    "(artículo 31 del Reglamento de La Torre)."
                )

            # Evaluar cantidad solamente cuando se habla de invitados
            if "invitados" in palabras or "invitado" in palabras:

                cantidad = entidades["cantidad"]

                if cantidad is not None and cantidad > 6:
                    restricciones.append(
                        "Se supera el máximo permitido de seis "
                        "invitados por lote (artículo 34 del "
                        "Reglamento de La Torre)."
                    )

        return restricciones

    def manejar_aclaracion(self, mensaje):

            palabras = set(normalizar_texto(mensaje))

            if self.estado == "esperando_ubicacion_evento":

                if "terraza" in palabras:
                    resultado = self.base.obtener_respuesta(
                      "consultar_reservaciones",
                      "terraza"
                    )
                    self.estado = None
                    if resultado is None:
                      return "No encontre informacion sobre reservaciones."
                    return (
                        resultado["respuesta"]
                        + "\nFuente: "
                        + resultado["fuente"]
                    )

                elif "casa" in palabras or "vivienda" in palabras:
                    self.estado = None
                return (
                    "Para eventos dentro de una vivienda particular "
                    "debes respetar las disposiciones del reglamento "
                    "general. Si necesitas conocer restricciones "
                    "específicas, consulta con la administración."
                  )
            else:
                return (
                    "¿El evento será en la terraza del coto "
                    "o dentro de tu vivienda?"
                )
            return None

    def responder(self, mensaje):

        # Paso 1: Comprobar si existe una aclaración pendiente
        if self.estado is not None:

            respuesta_aclaracion = self.manejar_aclaracion(mensaje)
            if respuesta_aclaracion is not None:
              return respuesta_aclaracion


        # Paso 2: identificar la intención
        intencion = self.detector.detectar_intencion(mensaje)

        if intencion == "intencion_desconocida":
            return (
                "Lo siento, no pude identificar tu consulta. "
                "Puedo ayudarte con reglamentos, horarios, "
                "áreas comunes, reservaciones, cuotas, "
                "construcción, mascotas y vialidades."
            )

        # Paso 3: Identificar consultas sobre eventos sin ubicación
        palabras = set(normalizar_texto(mensaje))

        if (
            any(palabra in palabras for palabra in
                ["fiesta", "evento", "celebracion"])
            and "terraza" not in palabras
            and "casa" not in palabras
            and "vivienda" not in palabras
        ):

            self.estado = "esperando_ubicacion_evento"

            return (
                "¿El evento será en la terraza del coto "
                "o dentro de tu vivienda?"
            )

        # Paso 4: Evaluar restricciones específicas
        restricciones = self.evaluar_restricciones(intencion, mensaje)

        if restricciones:
            return (
                "De acuerdo con el reglamento, existen "
                "las siguientes restricciones:\n"
                + "\n".join("- " + regla for regla in restricciones)
            )


        # Paso 5: identificar el tema
        tema = self.identificar_tema(intencion, mensaje)

        if tema is None:
            return (
                "Comprendo el tema de tu consulta, pero necesito "
                "más información para responder correctamente."
            )

        # Paso 6: recuperar información
        resultado = self.base.obtener_respuesta(intencion, tema)

        if resultado is None:
            return (
                "No cuento con información documentada "
                "para responder esta consulta."
            )

        # Paso 6: construir respuesta con su fuente
        return (
            resultado["respuesta"]
            + "\nFuente: "
            + resultado["fuente"]
        )

```
